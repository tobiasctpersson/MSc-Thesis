```python
import pandas as pd 
import numpy as np
file_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/Cleaned Data Sets/RV_total.csv"
df = pd.read_csv(file_path)

# Convert "Date" column to datetime 
df["Date"] = pd.to_datetime(df["Date"], utc=True).dt.tz_convert("Europe/Berlin")
df = df.sort_values("Date").set_index("Date")

df.tail()
```

## 1.1 HAR Model

HAR uses *heterogeneous horizons*: 
* **Daily** component: yesterday’s volatility
* **Weekly** component: average volatility over the last 5 trading days
* **Monthly** component: average volatility over the last 22 trading days

These will later enter into the simplest HAR equation: 
$$
RV_{t+1}
= \beta_0
+ \beta_d \, RV_t^{(d)}
+ \beta_w \, RV_t^{(w)}
+ \beta_m \, RV_t^{(m)}
+ \varepsilon_{t+1}
$$

where

- $RV_t^{(d)} = RV_t$
- $RV_t^{(w)} = \frac{1}{5} \sum_{i=0}^{4} RV_{t-i}$
- $RV_t^{(m)} = \frac{1}{22} \sum_{i=0}^{21} RV_{t-i}$

We will use "RV_total" as $RV_t$

```python
# Create dedicated dataframe for baseline HAR
df_HAR = df.sort_index().copy()

# Base RV series
rv = df_HAR["RV_total"]

# HAR regressors available at time t
df_HAR["RV_d"] = rv
df_HAR["RV_w"] = rv.rolling(window=5).mean()
df_HAR["RV_m"] = rv.rolling(window=22).mean()

# Forecast target (next day's RV)
df_HAR["RV_target"] = rv.shift(-1)
```

```python
# Keep only rows where all regressors and target are available
df_HAR = df_HAR.dropna(subset=["RV_d", "RV_w", "RV_m", "RV_target"])

print("First usable date:", df_HAR.index.min())
print("Last usable date:", df_HAR.index.max())
print("Number of usable observations:", len(df_HAR))
```

**Deciding on train, validation, and test splits:**
* Train = 1998 - 2012 (Train includes dot-com + GFC → rich volatility regimes) *~60–65% train*
* Validation = 2013 - 2015 (Validation includes Euro crisis aftershocks) *~15% validation*
* Test = 2016 - 2020 (Test includes calm + COVID shock) *~20–25% test*

```python
# Define split dates
train_end = "2012-12-31"
val_end   = "2015-12-31"

# Chronological split
df_train = df_HAR.loc[:train_end]
df_val   = df_HAR.loc[train_end:val_end]
df_test  = df_HAR.loc[val_end:]

# Remove overlap at boundary (since slicing is inclusive)
df_val   = df_val.iloc[1:]
df_test  = df_test.iloc[1:]

# Print sizes
print("Train observations:", len(df_train))
print("Validation observations:", len(df_val))
print("Test observations:", len(df_test))

print("\nTrain period:", df_train.index.min(), "to", df_train.index.max())
print("Validation period:", df_val.index.min(), "to", df_val.index.max())
print("Test period:", df_test.index.min(), "to", df_test.index.max())
```

#### HAR on Training set 

```python
import statsmodels.api as sm

# Define regressors and target (training only)
X_train = df_train[["RV_d", "RV_w", "RV_m"]]
y_train = df_train["RV_target"]

# Add intercept
X_train = sm.add_constant(X_train)

# Estimate OLS
har_model = sm.OLS(y_train, X_train).fit()

# Print summary
print(har_model.summary())
```

**Estimated model:** 
$$
RV_{t+1}
= 1.57 \times 10^{-5}
+ 0.363 \, RV_t
+ 0.366 \, RV_t^{(w)}
+ 0.179 \, RV_t^{(m)}
+ \varepsilon_{t+1}
$$

R-squared implies that about **58%** of next-day realized variance (RV) is explained by HAR structure (in-sample)

#### HAR on Rolling Estimation (Validation + Test)

```python
from tqdm import tqdm

# Combine validation and test into one evaluation window
df_eval = pd.concat([df_val, df_test])

predictions, actuals, dates = [], [], []

for current_date in tqdm(df_eval.index):

    # Expanding window: use all data strictly before current_date
    df_expanding = df_HAR.loc[:current_date].iloc[:-1]

    X = df_expanding[["RV_d", "RV_w", "RV_m"]]
    y = df_expanding["RV_target"]

    # Add intercept
    X = sm.add_constant(X, has_constant="add")

    model = sm.OLS(y, X).fit()

    # Forecast row as a 1-row DataFrame with matching columns
    X_fore = df_HAR.loc[[current_date], ["RV_d", "RV_w", "RV_m"]]
    X_fore = sm.add_constant(X_fore, has_constant="add")

    # Ensure column order matches training exactly
    X_fore = X_fore.reindex(columns=X.columns)

    forecast = float(model.predict(X_fore).iloc[0])

    predictions.append(forecast)
    actuals.append(float(df_HAR.loc[current_date, "RV_target"]))
    dates.append(current_date)

df_forecast = pd.DataFrame({"Forecast": predictions, "Actual": actuals}, index=dates)

print(df_forecast.head())
print("\nForecast df shape:", df_forecast.shape)
```

```python
# Split the forecast dataframe back into validation/test using the same boundaries as earlier
val_start, val_end = df_val.index.min(), df_val.index.max()
test_start, test_end = df_test.index.min(), df_test.index.max()

df_forecast_val = df_forecast.loc[val_start:val_end]
df_forecast_test = df_forecast.loc[test_start:test_end]

# MSE
mse_val = np.mean((df_forecast_val["Actual"] - df_forecast_val["Forecast"])**2)
mse_test = np.mean((df_forecast_test["Actual"] - df_forecast_test["Forecast"])**2)

print("Validation period:", val_start, "to", val_end, "| n =", len(df_forecast_val))
print("Test period:", test_start, "to", test_end, "| n =", len(df_forecast_test))

print("\nValidation MSE:", mse_val)
print("Test MSE:", mse_test)
```

Test MSE is double the size of Validation MSE, which is economically inuitive as the test period includes the 2018 volatility spike and the 2020 COVID shock. Volatility forecasting models struggle more during crisis regimes.

Hom much better HAR is than simply forecasting historical mean: 
$$
R^2_{\text{OOS}}
= 1
- \frac{\sum \left( RV_{t+1} - \widehat{RV}_{t+1} \right)^2}
       {\sum \left( RV_{t+1} - \overline{RV} \right)^2}
$$


```python
# Benchmark: mean RV from TRAINING sample only
mean_train = df_train["RV_target"].mean()

# --- Validation ---
sse_val = np.sum((df_forecast_val["Actual"] - df_forecast_val["Forecast"])**2)
sse_val_bench = np.sum((df_forecast_val["Actual"] - mean_train)**2)

r2_oos_val = 1 - (sse_val / sse_val_bench)

# --- Test ---
sse_test = np.sum((df_forecast_test["Actual"] - df_forecast_test["Forecast"])**2)
sse_test_bench = np.sum((df_forecast_test["Actual"] - mean_train)**2)

r2_oos_test = 1 - (sse_test / sse_test_bench)

print("Out-of-sample R^2 (Validation):", r2_oos_val)
print("Out-of-sample R^2 (Test):", r2_oos_test)
```

* $R^2_{\text{OOS}} > 0$ --> model beats mean forecast
* $R^2_{\text{OOS}} = 0$ --> same as mean forecast
* $R^2_{\text{OOS}} < 0$ --> wore than mean forecast
