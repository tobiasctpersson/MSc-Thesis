# Random Forest (RF) on the MHAR 5-min Feature Set


**Purpose:** The purpose of this notebook is to compute the out-of-sample forecast performance of the Random Forest (RF) model on the $\mathcal{M}_{\mathrm{HAR}}$ 5-min feature set for the EURO STOXX 50 index over the period 1999--2020, following the methodology of Christensen et al. (2023).

The Random Forest model builds on regression trees by averaging across a large number of bootstrapped trees. Relative to linear HAR-type models, RF is a fully nonparametric approach that can capture nonlinearities and interaction effects in the realized-volatility predictors without requiring them to be specified in advance.

The baseline forecasting relation is written as:

$$
RV_{t+1} = f(Z_t) + \varepsilon_{t+1},
$$

where $Z_t$ consists of the MHAR predictors:

- $RV_t$ (RVD) — daily realized variance  
- $\overline{RV}_t^{(w)} = \frac{1}{5}\sum_{i=0}^{4} RV_{t-i}$ (RVW) — 5-day rolling average of realized variance  
- $\overline{RV}_t^{(m)} = \frac{1}{22}\sum_{i=0}^{21} RV_{t-i}$ (RVM) — 22-day rolling average of realized variance  

The RF predictor is obtained by averaging over $B$ regression trees:

$$
\hat f(Z_t) = \frac{1}{B}\sum_{b=1}^{B} \hat f_b(Z_t),
$$

where each tree is estimated on a bootstrap sample and, at each split, only a random subset of predictors is considered. In line with the paper, RF is treated as an off-the-shelf tree-based model, so no separate hyperparameter tuning is performed. Instead, the forest is re-estimated in a fixed-length rolling window using the combined training and validation sample available at each forecast origin.


## 1. Loading Data
Loading in data from the `MALL_5min.csv` file.


```python
import pandas as pd
import numpy as np
import os
import time
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# Load MALL dataset
file_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/MALL_5min.csv"
MALL = pd.read_csv(file_path, index_col="Date", parse_dates=True)

print(f"Shape: {MALL.shape}")
print(f"Date range: {MALL.index[0].date()} to {MALL.index[-1].date()}")
print(f"NaN values: {MALL.isna().sum().sum()}")
MALL.head()
```

## 2. Data Split
### 2.1 Train / Validation / Test Split (70 / 10 / 20)


Following Christensen et al. (2023), we split the sample into a training set (70%), validation set (10%), and test set (20%). For the off-the-shelf RF implementation, the training and validation blocks are then combined into a single rolling estimation window used to generate one-day-ahead test-set forecasts.


```python
# Train / Validation / Test split (70 / 10 / 20)
n = len(MALL)
n_train = int(np.floor(0.70 * n))
n_val = int(np.floor(0.10 * n))
n_test = n - n_train - n_val

print(f"Training observations:   {n_train}")
print(f"Validation observations: {n_val}")
print(f"Test observations:       {n_test}")

print(f"Training period:   {MALL.index[0].date()} to {MALL.index[n_train - 1].date()}")
print(f"Validation period: {MALL.index[n_train].date()} to {MALL.index[n_train + n_val - 1].date()}")
print(f"Test period:       {MALL.index[n_train + n_val].date()} to {MALL.index[-1].date()}")
```

## 3. Random Forest
### 3.1 Model Specification


The Random Forest model replaces the linear HAR forecasting equation with an ensemble of regression trees. Since the predictors are only `RVD`, `RVW`, and `RVM`, the model remains deliberately compact, but it can still capture nonlinear threshold effects and interactions among the realized-volatility measures.


### 3.2 Feature Set Construction
Extracting the dependent and independent variables relevant for the RF model.


```python
# Define target and MHAR feature set
y = MALL["RV_target"].values
X_MHAR = MALL[["RVD", "RVW", "RVM"]].values

# Split into train, validation and test sets
X_train = X_MHAR[:n_train]
X_val = X_MHAR[n_train:n_train + n_val]
X_test = X_MHAR[n_train + n_val:]

y_train = y[:n_train]
y_val = y[n_train:n_train + n_val]
y_test = y[n_train + n_val:]

# Standardize predictors using training-sample moments for consistency across notebooks
X_mean = X_train.mean(axis=0)
X_std = X_train.std(axis=0, ddof=0)
X_std = np.where(X_std == 0, 1.0, X_std)

X_scaled = (X_MHAR - X_mean) / X_std
X_train_scaled = X_scaled[:n_train]
X_val_scaled = X_scaled[n_train:n_train + n_val]
X_test_scaled = X_scaled[n_train + n_val:]

print("Feature set: RVD, RVW, RVM")
print(f"\nFull dataset:   X={X_scaled.shape}, y={y.shape}")
print(f"Training set:   X={X_train_scaled.shape}, y={y_train.shape}")
print(f"Validation set: X={X_val_scaled.shape}, y={y_val.shape}")
print(f"Test set:       X={X_test_scaled.shape}, y={y_test.shape}")
print("\nStandardization check on training sample:")
print(f"  Means: {X_train_scaled.mean(axis=0).round(6)}")
print(f"  Std:   {X_train_scaled.std(axis=0).round(6)}")
```

## 5. Random Forest Model Specification

Following Christensen et al. (2023), RF is treated as an off-the-shelf tree-based model rather than a heavily tuned learner. The forest is therefore estimated with a fixed parameter configuration and then re-estimated across rolling windows over the test period.


```python
# Random Forest settings following the paper-style off-the-shelf implementation
n_features = X_scaled.shape[1]
max_features = max(1, n_features // 3)

rf_params = {
    "n_estimators": 500,
    "min_samples_leaf": 5,
    "max_features": max_features,
    "bootstrap": True,
    "random_state": 42,
    "n_jobs": 1
}

print("Random Forest parameters:")
for key, value in rf_params.items():
    print(f"  {key}: {value}")

rf_initial = RandomForestRegressor(**rf_params)
rf_initial.fit(X_scaled[:n_train + n_val], y[:n_train + n_val])

feature_importance = pd.Series(
    rf_initial.feature_importances_,
    index=["RVD", "RVW", "RVM"]
)

display(feature_importance.sort_values(ascending=False).to_frame("Initial feature importance"))
```

## 6. Rolling Window Estimation and Forecasting

Each forecast is formed one day ahead using a rolling estimation window of length equal to the initial training plus validation sample. Since RF is treated as an off-the-shelf learner, the forest is simply re-estimated in each window without separate hyperparameter tuning.


```python
window_size = n_train + n_val

print(f"Rolling window length (train + validation): {window_size}")

forecasts_RF = []
dates_test = []
importance_path = []

start_time = time.time()

for i in range(n_test):
    X_window = X_scaled[i:i + window_size]
    y_window = y[i:i + window_size]

    model_final = RandomForestRegressor(**rf_params)
    model_final.fit(X_window, y_window)

    X_forecast = X_scaled[i + window_size].reshape(1, -1)
    forecast = model_final.predict(X_forecast)[0]

    # Paper-consistent floor for negative volatility forecasts
    min_rv_roll = y_window.min()
    forecast = max(forecast, min_rv_roll)

    forecasts_RF.append(forecast)
    dates_test.append(MALL.index[i + window_size])
    importance_path.append(model_final.feature_importances_)

    if (i + 1) % 100 == 0:
        elapsed = (time.time() - start_time) / 60
        print(f"  Day {i + 1:4d}/{n_test} | Date: {MALL.index[i + window_size].date()} | Elapsed: {elapsed:.1f} min")

total_elapsed = (time.time() - start_time) / 60
print(f"\nTotal estimation time: {total_elapsed:.1f} minutes")
print(f"Forecasts generated: {len(forecasts_RF)}")
print(f"Forecast period: {dates_test[0].date()} to {dates_test[-1].date()}")

avg_importance = pd.Series(
    np.mean(np.vstack(importance_path), axis=0),
    index=["RVD", "RVW", "RVM"]
)
print("\nAverage feature importance across rolling windows:")
display(avg_importance.sort_values(ascending=False).to_frame("Average importance"))
```

## 7. Save Forecasts

The forecasts and summary metrics are stored using the same file structure as the benchmark and regularized-model notebooks so they can be reused in downstream model-comparison work.


```python
import os

# --- Define paths ---
forecast_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/Forecasts 5-min/MHAR/Tree-based Models"
mse_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/MSE 5-min/MHAR/Tree-based Models"

os.makedirs(forecast_path, exist_ok=True)
os.makedirs(mse_path, exist_ok=True)

# --- Construct results dataframe ---
results_RF = pd.DataFrame({
    "RV_actual": y_test,
    "RV_forecast": forecasts_RF
}, index=dates_test)

# --- Compute errors ---
errors_RF = results_RF["RV_actual"] - results_RF["RV_forecast"]
mse_RF = np.mean(errors_RF ** 2)

# --- Save forecasts ---
results_RF.to_csv(os.path.join(forecast_path, "forecasts_RF.csv"))

# --- Save MSE summary ---
mse_summary_RF = pd.DataFrame({
    "Model": ["RF"],
    "MSE": [mse_RF],
    "RMSE": [np.sqrt(mse_RF)],
    "MAE": [errors_RF.abs().mean()],
    "Mean_Error": [errors_RF.mean()],
}).set_index("Model")

mse_summary_RF.to_csv(os.path.join(mse_path, "mse_summary_RF.csv"))

# --- Print summary ---
print(f"Forecasts saved to: {forecast_path}")
print(f"MSE summary saved to: {mse_path}")
print(f"Rows saved: {len(results_RF)}")
print(f"Random Forest test-set MSE: {mse_RF:.6e}")
```

## 4. Results
### 4.1 Forecast Visualization


```python
import matplotlib.pyplot as plt

# Convert to annualized volatility for readability
results_plot = pd.DataFrame({
    "Actual": np.sqrt(results_RF["RV_actual"] * 252) * 100,
    "Forecast": np.sqrt(results_RF["RV_forecast"] * 252) * 100
}, index=results_RF.index)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(results_plot["Actual"], label="Actual RV", color="black", linewidth=0.8)
ax.plot(results_plot["Forecast"], label="Random Forest Forecast", color="red", linewidth=0.8, linestyle="--")
ax.set_title("Random Forest — One-Day-Ahead Realized Volatility Forecast\nEURO STOXX 50 (2016–2020)", fontsize=12)
ax.set_ylabel("Annualized Volatility (%)")
ax.set_xlabel("Date")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nSummary of forecast errors:")
errors = results_RF["RV_actual"] - results_RF["RV_forecast"]
print(f"Mean error:  {errors.mean():.6e}")
print(f"RMSE:        {np.sqrt(mse_RF):.6e}")
print(f"MAE:         {errors.abs().mean():.6e}")
```
