# HAR-X Model on the MHAR 5-min Feature Set


**Purpose:** The purpose of this notebook is to compute the out-of-sample forecast performance of the Heterogeneous Autoregressive model with exogenous predictors (HAR-X) on the $\mathcal{M}_{\mathrm{HAR}}$ 5-min feature set for the EURO STOXX 50 index over the period 1999--2020, following the methodology of Christensen et al. (2023).

On the $\mathcal{M}_{\mathrm{HAR}}$ feature set, the HAR-X specification collapses to the standard HAR model because the information set contains only the realized-variance block and no additional exogenous predictors. Therefore, the forecasts and test-set MSE from this notebook should match those of the plain HAR notebook exactly, provided the same rolling-window OLS procedure is used.

The model is defined as:

$$RV_{t+1} = \beta_0 + \beta' Z_t + \varepsilon_{t+1},$$

where $Z_t = (RVD_t, RVW_t, RVM_t)'$ for the MHAR feature set. Equivalently, this is the standard HAR regression:

$$RV_{t+1} = \beta_0 + \beta_d RV_t + \beta_w \overline{RV}_t^{(w)} + \beta_m \overline{RV}_t^{(m)} + \varepsilon_{t+1}.$$

The model is estimated via Ordinary Least Squares (OLS) using a rolling window of fixed length equal to the combined training and validation set, rolling forward one day at a time across the test period. In line with the benchmark-model implementation used elsewhere in the thesis, negative forecasts are replaced by the minimum in-sample realized variance.


## 1. Loading Data
Loading in data from the MALL_5min.csv file

```python
import pandas as pd
import numpy as np

# Load MALL dataset
file_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/MALL_5min.csv"
MALL = pd.read_csv(file_path, index_col='Date', parse_dates=True)

print(f"Shape: {MALL.shape}")
print(f"Date range: {MALL.index[0].date()} to {MALL.index[-1].date()}")
print(f"NaN values: {MALL.isna().sum().sum()}")
MALL.head()
```

## 2. Data Split
### 2.1 Train / Validation / Test Split (70 / 10 / 20)

Following Christensen et al. (2023), we split the sample into a training set (70%), 
validation set (10%), and test set (20%). The test set is reserved exclusively for 
out-of-sample evaluation and no estimation or tuning is conducted on it.

```python
# Train / Validation / Test split (70 / 10 / 20)
n = len(MALL)
n_train = int(np.floor(0.70 * n))
n_val   = int(np.floor(0.10 * n))
n_test  = n - n_train - n_val

train_idx = MALL.index[:n_train]
val_idx   = MALL.index[n_train:n_train + n_val]
test_idx  = MALL.index[n_train + n_val:]

print(f"Total observations: {n}")
print(f"\nTrain:      {train_idx[0].date()} to {train_idx[-1].date()} ({n_train} obs, {n_train/n*100:.1f}%)")
print(f"Validation: {val_idx[0].date()} to {val_idx[-1].date()} ({n_val} obs, {n_val/n*100:.1f}%)")
print(f"Test:       {test_idx[0].date()} to {test_idx[-1].date()} ({n_test} obs, {n_test/n*100:.1f}%)")
```

## 3. HAR-X Model
### 3.1 Model Specification


The HAR-X specification estimated in this notebook is:

$$RV_{t+1} = \beta_0 + \beta' Z_t + \varepsilon_{t+1},$$

where the MHAR information vector is

$$Z_t = (RVD_t, RVW_t, RVM_t)'$$

Because no additional exogenous predictors are present in the MHAR feature set, this HAR-X specification is numerically identical to the standard HAR benchmark.


### 3.2 Feature Set Construction
Extracting the dependent and independent variables relevant for the HAR-X model.


```python
# Define target and feature set for MHAR HAR-X
y = MALL["RV_target"]
X_HARX = MALL[["RVD", "RVW", "RVM"]]

print(f"Target variable: RV_target")
print(f"Predictors: {X_HARX.columns.tolist()}")
print(f"\nFirst few rows:")
print(pd.concat([y, X_HARX], axis=1).head())
```

### 3.3 Rolling Window Estimation and Forecasting

Following Christensen et al. (2023), the HAR-X model is estimated via OLS using a rolling window of fixed length equal to the combined training and validation set. The window rolls forward one day at a time across the test set, producing one-step-ahead forecasts. Since HAR-X on MHAR uses only `RVD`, `RVW`, and `RVM`, the resulting forecasts should coincide with those from the HAR benchmark. Negative forecasts are replaced by the minimum in-sample realized variance following the insanity filter of Bollerslev, Patton, and Quaedvlieg (2016).


```python
from sklearn.linear_model import LinearRegression

# Rolling window length = train + validation
window = n_train + n_val

forecasts_HARX = []
dates_test    = []

for i in range(n_test):
    # Extract rolling window
    X_window = X_HARX.iloc[i : i + window].values
    y_window = y.iloc[i : i + window].values
    
    # OLS estimation
    model = LinearRegression()
    model.fit(X_window, y_window)
    
    # One-step-ahead forecast using time t predictors
    X_forecast = X_HARX.iloc[i + window].values.reshape(1, -1)
    forecast   = model.predict(X_forecast)[0]
    
    # Insanity filter: replace negative forecasts with minimum in-sample RV
    min_rv   = y_window.min()
    forecast = max(forecast, min_rv)
    
    forecasts_HARX.append(forecast)
    dates_test.append(MALL.index[i + window])

print(f"Forecasts generated: {len(forecasts_HARX)}")
print(f"Forecast period: {dates_test[0].date()} to {dates_test[-1].date()}")
```

### 3.4 Out-of-Sample Performance Evaluation


Following Christensen et al. (2023), out-of-sample forecast accuracy is evaluated using the Mean Squared Error (MSE):

$$MSE = \frac{1}{T}\sum_{t \in \text{test}} (RV_{t+1} - \hat{f}(Z_t))^2$$

where $\hat{f}(Z_t)$ is the one-step-ahead forecast from the HAR-X model and $T$ is the number of test observations. On the MHAR feature set, this MSE should be identical to that of the HAR notebook.


```python
# Collect results
results_HARX = pd.DataFrame({
    "RV_actual": y.loc[dates_test].values,
    "RV_forecast": forecasts_HARX
}, index=dates_test)

# Compute out-of-sample MSE
mse_HARX = np.mean((results_HARX["RV_actual"] - results_HARX["RV_forecast"])**2)

print(f"Out-of-sample MSE (HARX): {mse_HARX:.6e}")
print(f"\nFirst few forecasts vs actuals:")
print(results_HARX.head(10).to_string())
```

### 3.4.1 Save Forecasts


```python
import os

# Paths
forecast_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/Forecasts 5-min/MHAR/Benchmark Models"
mse_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/MSE 5-min/MHAR/Benchmark Models"

os.makedirs(forecast_path, exist_ok=True)
os.makedirs(mse_path, exist_ok=True)

results_HARX.to_csv(os.path.join(forecast_path, "forecasts_HARX.csv"))

mse_summary_HARX = pd.DataFrame({
    "Model": ["HARX"],
    "MSE": [mse_HARX],
    "RMSE": [np.sqrt(mse_HARX)],
    "MAE": [np.mean(np.abs(results_HARX["RV_actual"] - results_HARX["RV_forecast"]))],
    "Mean_Error": [np.mean(results_HARX["RV_actual"] - results_HARX["RV_forecast"])]
}).set_index("Model")

mse_summary_HARX.to_csv(os.path.join(mse_path, "mse_summary_HARX.csv"))

print(f"Forecasts saved to: {forecast_path}")
print(f"MSE saved to: {mse_path}")
print(f"Files saved:")
print(f"  - forecasts_HARX.csv ({len(results_HARX)} rows)")
print(f"  - mse_summary_HARX.csv")
print(f"\nHARX MSE: {mse_HARX:.6e}")
```

## 4. Results
### 4.1 Forecast Visualization


```python
import matplotlib.pyplot as plt

# Convert to annualized volatility for readability
results_plot = pd.DataFrame({
    "Actual": np.sqrt(results_HARX["RV_actual"] * 252) * 100,
    "Forecast": np.sqrt(results_HARX["RV_forecast"] * 252) * 100
}, index=results_HARX.index)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(results_plot["Actual"], label="Actual RV", color="black", linewidth=0.8)
ax.plot(results_plot["Forecast"], label="HARX Forecast", color="red", linewidth=0.8, linestyle="--")
ax.set_title("HAR-X Model — One-Day-Ahead Realized Volatility Forecast\nEURO STOXX 50 (2016–2020)", fontsize=12)
ax.set_ylabel("Annualized Volatility (%)")
ax.set_xlabel("Date")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nSummary of forecast errors:")
errors = results_HARX["RV_actual"] - results_HARX["RV_forecast"]
print(f"Mean error:  {errors.mean():.6e}")
print(f"RMSE:        {np.sqrt(mse_HARX):.6e}")
print(f"MAE:         {errors.abs().mean():.6e}")
```
