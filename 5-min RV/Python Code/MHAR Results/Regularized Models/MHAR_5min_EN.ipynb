# Elastic Net (EN) on the MHAR 5-min Feature Set

**Purpose:** The purpose of this notebook is to compute the out-of-sample forecast performance of the Elastic Net regression model on the $\mathcal{M}_{\mathrm{HAR}}$ 5-min feature set for the EURO STOXX 50 index over the period 1999--2020, following the methodology of  
Christensen et al. (2023).

The Elastic Net model builds on the standard HAR specification by introducing a combination of $L_1$- and $L_2$-penalties on the slope coefficients. This allows the model to perform both shrinkage and variable selection, while also stabilizing coefficient estimates in the presence of multicollinearity. In particular, the Elastic Net can be seen as a compromise between ridge regression and LASSO, retaining the grouping effect of ridge while allowing for sparsity as in LASSO.

The model is defined as:

$$
RV_{t+1} = \beta_0 + \beta_d RV_t + \beta_w \overline{RV}_t^{(w)} + \beta_m \overline{RV}_t^{(m)} + \varepsilon_{t+1}
$$

where the predictors are the current day's realized variance measures:

- $RV_t$ (RVD) — daily realized variance  
- $\overline{RV}_t^{(w)} = \frac{1}{5}\sum_{i=0}^{4} RV_{t-i}$ (RVW) — 5-day rolling average of realized variance  
- $\overline{RV}_t^{(m)} = \frac{1}{22}\sum_{i=0}^{21} RV_{t-i}$ (RVM) — 22-day rolling average of realized variance  

The coefficients are estimated by solving the regularized least squares problem:

$$
\min_{\beta_0,\beta}
\left\{
\sum_{t=1}^{T}
\left(RV_{t+1} - \beta_0 - X_t \beta\right)^2
+ \lambda \left[
\alpha \sum_{j=1}^{p} |\beta_j|
+ (1 - \alpha) \sum_{j=1}^{p} \beta_j^2
\right]
\right\},
$$

where $\lambda \geq 0$ controls the overall degree of regularization and $\alpha \in [0,1]$ determines the relative weight between the $L_1$ and $L_2$ penalties. Setting $\alpha = 1$ corresponds to LASSO, while $\alpha = 0$ corresponds to ridge regression.

The model is estimated using a rolling forecasting scheme with a fixed train/validation/test split (70/10/20). At each step, the optimal combination of regularization parameters is selected based on validation set performance, and a one-day-ahead forecast is generated throughout the test period to evaluate out-of-sample accuracy.

## 1. Loading Data
Loading in data from the MALL_5min.csv file

```python
import pandas as pd
import numpy as np
import os
import time
import matplotlib.pyplot as plt
from sklearn.linear_model import ElasticNet

# Load MALL dataset
file_path = "/Users/tobiasbergdahlpersson/Documents/SSE/MSc Thesis/5-min RV/Cleaned Data Sets/MALL_5min.csv"
MALL = pd.read_csv(file_path, index_col='Date', parse_dates=True)

print(f"Shape: {MALL.shape}")
print(f"Date range: {MALL.index[0].date()} to {MALL.index[-1].date()}")
print(f"NaN values: {MALL.isna().sum().sum()}")
MALL.head()
```

## 2. Data Split
### 2.1 Train / Validation / Test Split (70 / 10 / 20)

Following Christensen et al. (2023), we split the sample into a training set (70%), 
validation set (10%), and test set (20%). The test set is reserved exclusively for 
out-of-sample evaluation and no estimation or tuning is conducted on it.

```python
# Train / Validation / Test split (70 / 10 / 20)
n = len(MALL)
n_train = int(np.floor(0.70 * n))
n_val   = int(np.floor(0.10 * n))
n_test  = n - n_train - n_val

train_idx = MALL.index[:n_train]
val_idx   = MALL.index[n_train:n_train + n_val]
test_idx  = MALL.index[n_train + n_val:]

print(f"Total observations: {n}")
print(f"\nTrain:      {train_idx[0].date()} to {train_idx[-1].date()} ({n_train} obs, {n_train/n*100:.1f}%)")
print(f"Validation: {val_idx[0].date()} to {val_idx[-1].date()} ({n_val} obs, {n_val/n*100:.1f}%)")
print(f"Test:       {test_idx[0].date()} to {test_idx[-1].date()} ({n_test} obs, {n_test/n*100:.1f}%)")
```

## 3. Elastic Net
### 3.1 Model Specification

The Elastic Net regression model builds on the baseline HAR specification by introducing a combination of $L_1$- and $L_2$-penalties on the slope coefficients. The model is defined as:

$$
RV_{t+1} = \beta_0 + \beta_d RV_t + \beta_w \overline{RV}_t^{(w)} + \beta_m \overline{RV}_t^{(m)} + \varepsilon_{t+1}
$$

where the coefficients are estimated by solving the regularized least squares problem:

$$
\min_{\beta_0,\beta}
\left\{
\sum_{t=1}^{T}
\left(RV_{t+1} - \beta_0 - X_t \beta\right)^2
+ \lambda \left[
\alpha \sum_{j=1}^{p} |\beta_j|
+ (1 - \alpha) \sum_{j=1}^{p} \beta_j^2
\right]
\right\},
$$

with $\lambda \geq 0$ denoting the overall regularization parameter and $\alpha \in [0,1]$ controlling the relative weight between the $L_1$ and $L_2$ penalties. When $\alpha = 1$, the Elastic Net reduces to the LASSO model, while $\alpha = 0$ corresponds to ridge regression.

### 3.2 Feature Set Construction
Extracting the dependent and independent variables relevant for the EN model

```python
# Define target and MHAR feature set
y = MALL["RV_target"].values
X_MHAR = MALL[["RVD", "RVW", "RVM"]].values

# Split indices only — no standardization here
# Standardization will be applied inside the rolling window loop
# using only the training portion of each window

y_train = y[:n_train]
y_val   = y[n_train:n_train + n_val]
y_test  = y[n_train + n_val:]

print("Feature set: RVD, RVW, RVM")
print(f"\nTraining set:   X={X_MHAR[:n_train].shape}, y={y_train.shape}")
print(f"Validation set: X={X_MHAR[n_train:n_train+n_val].shape}, y={y_val.shape}")
print(f"Test set:       X={X_MHAR[n_train+n_val:].shape}, y={y_test.shape}")
print(f"\nTarget mean (train): {y_train.mean():.6e}")
print(f"Target std  (train): {y_train.std(ddof=0):.6e}")
print(f"Target range:        [{y_train.min():.6e}, {y_train.max():.6e}]")
```

## Step 5: Rolling Window Estimation with Window-Level Standardization

Following Christensen et al. (2023), Elastic Net is estimated using a rolling window 
scheme with joint hyperparameter tuning over both the regularization strength alpha and 
the L1/L2 mixing parameter l1_ratio. At each forecast origin, both features and the 
target are standardized using moments from the training portion of the current window 
only. The alpha grid spans [1e-8, 1e1] with 1000 points in log space, and l1_ratio is 
tuned over 10 values spanning [0.01, 1.00], yielding 10,000 combinations per window. 
The combination minimizing validation MSE is selected, the model is refit on the full 
training window, and the forecast is back-transformed to the original scale.

```python
from sklearn.linear_model import ElasticNet

# Hyperparameter grids
alpha_grid    = np.logspace(-8, 1, 1000)
l1_ratio_grid = np.array([0.01, 0.05, 0.10, 0.20, 0.30, 
                           0.40, 0.50, 0.70, 0.90, 1.00])

window_size  = n_train + n_val
forecasts_EN = []
dates_test   = []
best_alphas  = []
best_l1s     = []
nonzero_path = []

start_time = time.time()

for i in range(n_test):
    # --- Define rolling window ---
    X_window = X_MHAR[i:i + window_size]
    y_window = y[i:i + window_size]

    X_win_train = X_window[:n_train]
    X_win_val   = X_window[n_train:]
    y_win_train = y_window[:n_train]
    y_win_val   = y_window[n_train:]

    # --- Window-level feature standardization ---
    X_mean_w = X_win_train.mean(axis=0)
    X_std_w  = X_win_train.std(axis=0, ddof=0)
    X_std_w  = np.where(X_std_w == 0, 1.0, X_std_w)

    X_win_train_s = (X_win_train - X_mean_w) / X_std_w
    X_win_val_s   = (X_win_val   - X_mean_w) / X_std_w

    # --- Window-level target standardization ---
    y_mean_w = y_win_train.mean()
    y_std_w  = y_win_train.std(ddof=0)

    y_win_train_s = (y_win_train - y_mean_w) / y_std_w
    y_win_val_s   = (y_win_val   - y_mean_w) / y_std_w

    # --- Forecast point ---
    X_fore   = X_MHAR[i + window_size].reshape(1, -1)
    X_fore_s = (X_fore - X_mean_w) / X_std_w

    # --- Joint 2D hyperparameter tuning ---
    best_mse_w   = np.inf
    best_alpha_w = None
    best_l1_w    = None

    for l1_ratio in l1_ratio_grid:
        for alpha in alpha_grid:
            model = ElasticNet(
                alpha=alpha,
                l1_ratio=l1_ratio,
                fit_intercept=True,
                max_iter=100000
            )
            model.fit(X_win_train_s, y_win_train_s)
            y_pred_val_s = model.predict(X_win_val_s)
            mse_w = np.mean((y_win_val_s - y_pred_val_s) ** 2)

            if mse_w < best_mse_w:
                best_mse_w   = mse_w
                best_alpha_w = alpha
                best_l1_w    = l1_ratio

    # --- Refit on full training window with best params ---
    final_model = ElasticNet(
        alpha=best_alpha_w,
        l1_ratio=best_l1_w,
        fit_intercept=True,
        max_iter=100000
    )
    final_model.fit(X_win_train_s, y_win_train_s)

    # --- Sparsity tracking ---
    n_nonzero = np.sum(np.abs(final_model.coef_) > 1e-12)
    nonzero_path.append(n_nonzero)

    # --- Generate forecast and back-transform ---
    forecast_s    = final_model.predict(X_fore_s)[0]
    forecast_orig = forecast_s * y_std_w + y_mean_w
    forecast_orig = max(forecast_orig, 0.0)

    forecasts_EN.append(forecast_orig)
    dates_test.append(MALL.index[i + window_size])
    best_alphas.append(best_alpha_w)
    best_l1s.append(best_l1_w)

    if (i + 1) % 100 == 0:
        elapsed = (time.time() - start_time) / 60
        print(f"  Day {i+1:4d}/{n_test} | "
              f"Date: {MALL.index[i + window_size].date()} | "
              f"Best alpha: {best_alpha_w:.2e} | "
              f"Best l1_ratio: {best_l1_w:.2f} | "
              f"Non-zero: {n_nonzero}/3 | "
              f"Elapsed: {elapsed:.1f} min")

total_elapsed = (time.time() - start_time) / 60
print(f"\nTotal estimation time: {total_elapsed:.1f} minutes")
print(f"Forecasts generated:   {len(forecasts_EN)}")
print(f"Forecast period:       {dates_test[0].date()} to {dates_test[-1].date()}")

print(f"\nHyperparameter path summary:")
alpha_series = pd.Series(best_alphas)
l1_series    = pd.Series(best_l1s)
print(f"  Most common alpha:    {alpha_series.mode()[0]:.2e}")
print(f"  Most common l1_ratio: {l1_series.mode()[0]:.2f}")
print(f"  Min alpha used:       {alpha_series.min():.2e}")
print(f"  Max alpha used:       {alpha_series.max():.2e}")
print(f"  Boundary hits (min):  {(alpha_series <= alpha_grid[0]  * 1.001).sum()}")
print(f"  Boundary hits (max):  {(alpha_series >= alpha_grid[-1] * 0.999).sum()}")

print(f"\nSparsity summary:")
nonzero_series = pd.Series(nonzero_path)
print(f"  Mean non-zero coefs:  {nonzero_series.mean():.2f} / 3")
print(f"  All coefs zero:       {(nonzero_series == 0).sum()} windows")
print(f"  All coefs nonzero:    {(nonzero_series == 3).sum()} windows")
print(f"  Partial sparsity:     {((nonzero_series > 0) & (nonzero_series < 3)).sum()} windows")

print(f"\nl1_ratio distribution:")
print(l1_series.value_counts().sort_index().to_string())
```

## 6. Rolling Window Estimation and Forecasting

We now carry the best validation-set hyperparameters into the rolling forecast exercise. Each forecast is formed one day ahead using a window of length equal to the initial training plus validation sample, matching the benchmark design used throughout the thesis notebooks.

```python
alpha_grid = np.logspace(-5, 2, 1000)
l1_ratio_grid = np.array([0.01, 0.05, 0.10, 0.20, 0.30, 0.40, 0.50, 0.70, 0.90, 1.00])

forecasts_EN = []
dates_test = []
nonzero_path = []
alpha_path = []
l1_ratio_path = []

start_time = time.time()

for i in range(n_test):
    # Expanding training window up to the current forecast origin
    train_end = n_train + i
    val_end = train_end + n_val

    X_train_roll = X_MHAR[:train_end]
    y_train_roll = y[:train_end]

    X_val_roll = X_MHAR[train_end:val_end]
    y_val_roll = y[train_end:val_end]

    X_test_point = X_MHAR[val_end].reshape(1, -1)

    # Standardize using the rolling training sample only
    X_mean_roll = X_train_roll.mean(axis=0)
    X_std_roll = X_train_roll.std(axis=0, ddof=0)
    X_std_roll = np.where(X_std_roll == 0, 1.0, X_std_roll)

    X_train_roll_scaled = (X_train_roll - X_mean_roll) / X_std_roll
    X_val_roll_scaled = (X_val_roll - X_mean_roll) / X_std_roll
    X_test_point_scaled = (X_test_point - X_mean_roll) / X_std_roll

    # Tune alpha and l1_ratio on the rolling validation set
    best_mse_roll = np.inf
    best_alpha_roll = None
    best_l1_ratio_roll = None

    for l1_ratio in l1_ratio_grid:
        for alpha in alpha_grid:
            model = ElasticNet(
                alpha=alpha,
                l1_ratio=l1_ratio,
                fit_intercept=True,
                max_iter=100000
            )
            model.fit(X_train_roll_scaled, y_train_roll)
            y_val_pred = model.predict(X_val_roll_scaled)
            mse = np.mean((y_val_roll - y_val_pred) ** 2)

            if mse < best_mse_roll:
                best_mse_roll = mse
                best_alpha_roll = alpha
                best_l1_ratio_roll = l1_ratio

    # Refit on rolling training sample and forecast one day ahead
    model_final = ElasticNet(
        alpha=best_alpha_roll,
        l1_ratio=best_l1_ratio_roll,
        fit_intercept=True,
        max_iter=100000
    )
    model_final.fit(X_train_roll_scaled, y_train_roll)
    forecast = model_final.predict(X_test_point_scaled)[0]

    # Realized variance should not be negative
    forecast = max(forecast, 0.0)

    forecasts_EN.append(forecast)
    dates_test.append(MALL.index[val_end])
    nonzero_path.append(np.sum(np.abs(model_final.coef_) > 1e-12))
    alpha_path.append(best_alpha_roll)
    l1_ratio_path.append(best_l1_ratio_roll)

    if (i + 1) % 100 == 0:
        elapsed = (time.time() - start_time) / 60
        print(f"  Day {i + 1:4d}/{n_test} | Date: {MALL.index[val_end].date()} | Elapsed: {elapsed:.1f} min")

total_elapsed = (time.time() - start_time) / 60
print(f"\nTotal estimation time: {total_elapsed:.1f} minutes")
print(f"Forecasts generated: {len(forecasts_EN)}")
print(f"Forecast period: {dates_test[0].date()} to {dates_test[-1].date()}")
print(f"Average number of active predictors: {np.mean(nonzero_path):.2f}")
print(f"Average chosen alpha: {np.mean(alpha_path):.6e}")
print(f"Average chosen l1_ratio: {np.mean(l1_ratio_path):.4f}")
```
